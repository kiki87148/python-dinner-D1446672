把餐費與人數從文字轉成數字，依原始餐費加收 10% 服務費，再扣折價券，最後平均分攤。每人預算為 300 元。

執行前先想想：如果只多一個人分攤，整桌應付的總額會改變嗎？

## 測試一

分攤人數為 4 人，折價券為 100 元。

```python
# 原始資料
event_name = "  WEEKEND DINNER  "
meal_amount = " NT$1,200 "
people = "4"

service_rate = 0.10
discount = 100
budget_per_person = 300

# 保留原始字串，將整理後的資料存入新變數
clean_name = event_name.strip().lower()
amount_text = meal_amount.replace("NT$", "").replace(",", "").replace(" ", "")
meal_cost = float(amount_text)
people_count = int(people)

# 服務費按原價計算，折價券從加收服務費後的金額扣除
service_fee = meal_cost * service_rate
total_due = meal_cost + service_fee - discount
per_person = total_due / people_count
over_budget = per_person > budget_per_person

# .2f 讓金額顯示到小數點後兩位
print(f"聚餐名稱：{clean_name}")
print(f"原始餐費：{meal_cost:.2f} 元")
print(f"服務費：{service_fee:.2f} 元")
print(f"折價券：{discount:.2f} 元")
print(f"應付總額：{total_due:.2f} 元")
print(f"分攤人數：{people_count} 人")
print(f"每人費用：{per_person:.2f} 元")
print(f"每人預算：{budget_per_person:.2f} 元")
print(f"是否超過預算：{over_budget}")
```
